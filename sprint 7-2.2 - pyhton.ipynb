Exercici 2: Classificació de clients de segurs

    Treballes com a analista de dades en una companyia asseguradora.

    L'empresa vol calcular el risc dels clients, el preu estimat de la pòlissa i la probabilitat de frau, utilitzant un codi de comportament que ha estat generat per un model intern.



```python
# Tens la següent llista de clients:
# lista de diccionarios con los clientes y sus datos
clients = [ 
    {"nom": "Ana", "edat": 42, "codi": 0}, 
    {"nom": "Carlos", "edat": 30, "codi": 2}, 
    {"nom": "Isabel", "edat": 55, "codi": 3}, 
    {"nom": "Jorge", "edat": 40, "codi": 1}, 
    {"nom": "Marta", "edat": 28, "codi": 0} 
] 
                  
# I aquest diccionari amb les categories del sistema:
nivells_seguro = { 
    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01}, 
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05}, 
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15}, 
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30} 
} 

# El preu final s'ha de calcular amb la fórmula:
# preu_final = preu_base * (1 + edat/100)
```

```python
import pandas as pd
# 🎯 Objectius
#     » Classificar cada client utilitzant nivells_seguro (sense if/elif).
def classificar_clients_seguretat(clients):
    clients_dicc = []
    for client in clients:
        # print("client = ", client)
        client_processat = {
            "nom" : client["nom"].strip(),
            "edat" : client["edat"],
            "codi" : client["codi"],
            "categoria" : nivells_seguro[client["codi"]]["categoria"],
            "possibilitat_frau" : nivells_seguro[client["codi"]]["fraude"],
            "preu_base" : nivells_seguro[client["codi"]]["preu_base"],
            "preu_final" : round(nivells_seguro[client["codi"]]["preu_base"] * (1 + client["edat"]/100), 2)
        }
        # print("paciente_procesado = ", paciente_procesado)
        clients_dicc.append(client_processat)    
    # print(clients_dicc)
    return clients_dicc

def mostrar_dataframe(clients_processats):
    df = pd.DataFrame(clients_processats)
    # sort_values(by="fraude", ascending=False)
    return df.sort_values(by=["possibilitat_frau", "preu_final"], ascending=False)

def main():
    clients_processats = classificar_clients_seguretat(clients)
    return mostrar_dataframe(clients_processats)
main()
```
