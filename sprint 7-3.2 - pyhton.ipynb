Exercici 2: Processament automàtic de dades esportives

    Ara formes part d'un equip d'analítica dins una organització esportiva que gestiona dades de la lliga catalana. La teva companya de feina vol automatitzar la lectura del seu arxiu històric de resultats i extreure les estadístiques més importants. T'ha demanat fer un petit sistema en Python que processi el fitxer 📄 "historic partits.txt"

🎯 Objectius

Mitjançant un codi modular, clar i documentat, el teu programa ha de calcular:

    » 1.Quin equip és el més golejador.

    » 2.Quin és el més golejat.

    » 3.El nombre total de gols que ha fet cada equip.

    » 4.La classificació final ordenada per punts on cada victòria suma 3 pts, cada empat 1 pts y cada derrota 0 pts.

En aquest nivell no es dona cap funció. Ets tu qui ha de dissenyar un flux complet. El codi ha de reflectir una bona organització interna.

```python
noms_equips = [
    "Badalona", "Cerdanyola", "Cornellà", "Europa", "FC Barcelona",
    "Figueres", "Girona FC", "Granollers", "Llagostera", "Lleida Esportiu",
    "Manlleu", "Nàstic de Tarragona", "Olot", "Prat", "RCD Espanyol",
    "Reus Deportiu", "Sabadell", "Sant Andreu", "Terrassa", "Vilafranca"
]

equips = []
for nom in noms_equips:
    dades_equip = {
        "nom": nom,
        "gols_favor": 0,
        "gols_contra": 0,
        "guanyats": 0,
        "empatats": 0,
        "perduts": 0,
        "punts": 0
    }
    equips.append(dades_equip)

print(equips)
```

```python
def actualitzar_estadistica(nomLocal, nomVisitant, gols_favor_local, gols_contra_local, guanyatLocal, empatat, perdutLocal):
    # Loop to find the right dicc
    for equip in equips:
        if equip["nom"] == nomLocal:
            # print("equip['nomLocal'] = ", equip["nom"])
            equip["gols_favor"] += gols_favor_local
            equip["gols_contra"] += gols_contra_local
            equip["guanyats"] += guanyatLocal 
            equip["empatats"] += empatat
            equip["perduts"] += perdutLocal
            if guanyatLocal == 1:
                equip["punts"] += 3
            if empatat == 1:
                equip["punts"] += 1
    for equip in equips:
        if equip["nom"] == nomVisitant:
            # print("equip['nomVisitant'] = ", equip["nom"])
            equip["gols_favor"] += gols_contra_local  
            equip["gols_contra"] += gols_favor_local
            equip["guanyats"] += perdutLocal  
            equip["empatats"] += empatat
            equip["perduts"] += guanyatLocal
            if perdutLocal == 1:
                equip["punts"] += 3
            if empatat == 1:
                equip["punts"] += 1
    return equips
```

```python
def mostrar_datos(equips):
    # max_goleador = max(lambda)
    # Think of it as a two-step process happening in one line:
    max_goleador = max(equips, key=lambda x: x['gols_favor'])  
    mes_golejat = max(equips, key=lambda x: x['gols_contra'])
    classificacio = sorted(equips, key=lambda x: x['punts'], reverse=True)
    
    noms_equips = [x['nom'] for x in classificacio]
    print(f"Equip més golejador: {max_goleador['nom']}")
    print(f"Equip més golejat: {mes_golejat['nom']}")
    print("Classificació final:")
    for i, value in enumerate(classificacio, start=1):
        print(i , "." ,value["nom"])
        # print(f"...")
    
```

```python
import pandas as pd 

def mostrar_dataframe(equips_processats):
    df = pd.DataFrame(equips_processats)
    # sort_values(by="fraude", ascending=False)
    return df.sort_values(by=["punts"], ascending=False)
```

```python
def main():
    equips = []
    f = open("historic partits.txt", encoding="utf-8")
    line = f.readline()  
    while line != "":    # leer hasta q la ultima linea esté vacía
        # cada linea tiene sepracion por Tabs - local tab gol-gol tab visitante
        parts = (line.strip().split("\t"))
        nomLocal = parts[0]

        gols = parts[1].split("-")
        nomVisitant = parts[2]

        golsLocal = int(gols[0])
        golsVisitant = int(gols[1])
        if golsLocal > golsVisitant:
            # print("guanyador local")
            equips_added = actualitzar_estadistica(nomLocal, nomVisitant, golsLocal, golsVisitant, 1, 0, 0)
        elif golsVisitant > golsLocal:
            # print("guanyador visitant")
            equips_added = actualitzar_estadistica(nomLocal, nomVisitant, golsLocal, golsVisitant, 0, 0, 1)
        else: 
            equips_added = actualitzar_estadistica(nomLocal, nomVisitant, golsLocal, golsVisitant, 0, 1, 0)
        # print ("----------")
        line = f.readline()  
    f.close()
    # print("equips = ", equips_added)
    print(mostrar_datos(equips_added))

    return mostrar_dataframe(equips_added)
main()
#estefania thompson
```
