```python
def obtenir_dades():
    dades = {
        "nombre": input("Ingresa el nombre: "),
        "telefono": input("Ingresa tu teléfono: "),
        "experiencia": (input("Ingresa tus años de experiencia en la empresa: ")),
        "catalan": (input("""Conocimiento de Catalán
            0 = sense coneixement 
            1 = fins a nivell B1 
            2 = nivell B2 
            3 = C1 o superior 
        """)),
        "excel": (input("""Domini d'Excel 
            0 = nivell bàsic 
            1 = ús de funcions 
            2 = taules dinàmiques 
            3 = Visual Basic i Macros""")),
        "titulo": (input(""" 
            0 = ESO 
            1 = FP 
            2 = Universitat 
            3 = Màster o superior"""))
    }
    return dades
```

```python
#para evitar errores de ejecicion cuando se ponen campos en blanco e intentas convertir el string a int
def safe_int_convert(val, default=-1): #-1 returns false
    try:
        return int(val)
    except (ValueError, TypeError):
        return default
```

```python
def validar_dades(dades):
    # print("Validando datos...", dades)
    # Validar nombre
    if len(dades["nombre"].strip()) == 0:
        print("El nombre no puede estar vacío.")
        return False

    # Validar teléfono
    if dades["telefono"].isdigit() == False or len(dades["telefono"]) != 9:
        print("El teléfono debe ser un número de 9 dígitos.")
        return False

    # Validar años de experiencia
    if safe_int_convert(dades["experiencia"]) < 0:
        print("Los años de experiencia no pueden ser negativos.")
        return False

    # Validar nivel de catalán
    if safe_int_convert(dades["catalan"]) not in [0, 1, 2, 3]:
        print("El nivel de catalán debe ser 0, 1, 2 o 3.")
        return False

    # Validar dominio de Excel
    if safe_int_convert(dades["excel"]) not in [0, 1, 2, 3]:
        print("El dominio de Excel debe ser 0, 1, 2 o 3.")
        return False

    # Validar título académico
    if safe_int_convert(dades["titulo"]) not in [0, 1, 2, 3]:
        print("El título académico debe ser 0, 1, 2 o 3.")
        return False

    return True
```

```python
# El programa haurà de calcular una puntuació total aplicant el pes de l'experiència (anys * 1.5) 
# 'i sumant-hi els punts de català, excel i titulació, 
def calcular_puntuacio(dades):
    dades["puntuacio"] = (dades["experiencia"] * 1.5) + dades["catalan"] + dades["excel"] + dades["titulo"]
    return dades["puntuacio"]

```

```python
# a continuació classificar automàticament cada persona segons els llindars establerts per l'empresa:

#     » Prioritat Alta → puntuació ≥ 12
#     » Candidat adequat → 8 ≤ puntuació < 12
#     » No prioritari → puntuació < 8

def classificar_candidat(dades):
    if dades["puntuacio"] >= 12:
        dades["prioritat"] = "Prioritat Alta"
    elif 8 <= dades["puntuacio"] < 12:
        dades["prioritat"] = "Candidat adequat"
    else:
        dades["prioritat"] = "No prioritari"
    return dades["prioritat"]
```

```python
# Cada vegada que algú complet a el formulari, s'ha de mostrar la seva puntuació individual mostrar_individual
def mostrar_individual(dades):
    print(f"Nombre: {dades['nombre']}")
    print(f"Teléfono: {dades['telefono']}")
    print(f"Años de experiencia: {dades['experiencia']}")
    print(f"Nivel de catalán: {dades['catalan']}")
    print(f"Dominio de Excel: {dades['excel']}")
    print(f"Título académico: {dades['titulo']}")
    print(f"Puntuación total: {dades['puntuacio']}")
    print(f"Clasificación: {dades['prioritat']}")
    print("-" * 40)
```

```python
# Quan ja s'hagin inscrit exactament deu persones, l'eina ha de deixar d'acceptar nous registres i generar un informe final en pantalla, ordenat primer segons la prioritat i, en cas d'empat, per la puntuació obtinguda. Aquest informe ha d'incloure el nom, el telèfon i la puntuació de cada candidat per tal que el departament de RRHH els pugui contactar ràpidament.
def mostrar_informe_final(candidats_valids):
    # Ordenar por prioridad y puntuación
    prioritat_order = {"Prioritat Alta": 1, "Candidat adequat": 2, "No prioritari": 3}
    candidats_valids.sort(key=lambda x: (prioritat_order[x["prioritat"]], -x["puntuacio"]))

    print("\nInforme final de candidatos:")
    for num, candidat in enumerate(candidats_valids, start=1):
        print(f"{num}. Prioridad: {candidat['prioritat']}, Nombre: {candidat['nombre']}, Teléfono: {candidat['telefono']}, Puntuación: {candidat['puntuacio']}")
```

```python
def main():
    candidats_valids = []
    while len(candidats_valids) < 10:
        dades = obtenir_dades()
        # validar_dades(dades)
        if validar_dades(dades):
            calcular_puntuacio(dades)
            classificar_candidat(dades)
            mostrar_individual(dades)
            candidats_valids.append(dades)
            #asegurarse q solo acepta 10 candidats
            # if len(candidats_valids) == 10:
            #     break
        # print(dades)2211
    mostrar_informe_final(candidats_valids)
main()
```
