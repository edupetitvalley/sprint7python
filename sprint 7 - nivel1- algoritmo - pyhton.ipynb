- Función obtener datos(): 
    > no recibe ningún parámetro
    - crea un diccionario datos con las siguientes claves que obtienen los valores de sus inputs
        - nombre
        - telefono

        - catalan (niveles de 0 a 3)

```python
def obtener_datos():
    datos = {
        "nombre": input("Ingresa el nombre: "),
        "telefono": input("Ingresa tu teléfono: "),
        "experiencia": input("Ingresa tus años de experiencia en la empresa: "),
        "catalan": input("""Conocimiento de Catalán:
            0 = Sin conocimiento 
            1 = Hasta nivel B1 
            2 = Nivel B2 
            3 = C1 o superior 
        """),
        "excel": input("""Dominio de Excel:
            0 = Nivel básico 
            1 = Uso de funciones 
            2 = Tablas dinámicas 
            3 = Visual Basic y Macros
        """),
        "titulo": input("""Titulada académica:
            0 = ESO 
            1 = FP 
            2 = Universidad 
            3 = Máster o superior
        """)
    }
    return datos
```

- funcion validar_datos(datos): 
    > recibe el diccinario creado en obtener_datos
    - intenta:
        - delvolver la conversion de string a integer 
    - si no lo con- sigue:
        - imprime(el error)
        - retorna False para que interrumpir la funcion de validar_datos
    
    - si el campo nombre está vacío:
        - print("El nombre no puede estar vacío")
        - devuelve False para indicar que los datos no son válidos

    - si el campo telefono no es un dígito o si no tiene 9 digitos:
        - print("El teléfono debe ser un número de 9 dígitos")
        - devuelve False para indicar que los datos no son válidos
    
    - si despues de convertir a integer, el campo experiencia es negativo 
        - print("Los años de experiencia no pueden ser negativos.")
        - devuelve False para indicar que los datos no son válidos
    
    - si despues de convertir a integer, el campo catalan  no es 0, 1, 2 o 3:
        - print("El nivel de catalán debe ser 0, 1, 2 o 3.")
        - devuelve False para indicar que los datos no son válidos

    - si despues de convertir a integer,el campo excel no es 0, 1, 2 o 3:

        - devuelve False para indicar que los datos no son válidos

    - si despues de convertir a integer,el campo titulo no es 0, 1, 2, 3:

        - devuelve False para indicar que los datos no son válidos

    devuelve True si todas las condiciones de validacion han sido exitosas

```python
def validar_datos(datos):
    try:
        datos["experiencia"] = float((datos["experiencia"]))
        datos["catalan"] = int((datos["catalan"]))
        datos["excel"] = int((datos["excel"]))
        datos["titulo"] = int((datos["titulo"]))
        # return int(val)
    except Exception as e:
        print("error = ", e)
        return False
    
    # Validar nombre
    if len(datos["nombre"].strip()) == 0:
        print("El nombre no puede estar vacío.")
        return False

    # Validar teléfono
    if datos["telefono"].isdigit() == False or len(datos["telefono"]) != 9:
        print("El teléfono debe ser un número de 9 dígitos.")
        return False

    # Validar años de experiencia
    if (datos["experiencia"]) < 0:
        print("Los años de experiencia no pueden ser negativos.")
        return False

    # Validar nivel de catalán
    if (datos["catalan"]) not in [0, 1, 2, 3]:
        print("El nivel de catalán debe ser 0, 1, 2 o 3.")
        return False

    # Validar dominio de Excel
    if (datos["excel"]) not in [0, 1, 2, 3]:
        print("El dominio de Excel debe ser 0, 1, 2 o 3.")
        return False

    # Validar título académico
    if (datos["titulo"]) not in [0, 1, 2, 3]:
        print("El título académico debe ser 0, 1, 2 o 3.")
        return False

    return True
```

- funcion calcular_puntuacion(datos): 
    > recibe el diccionario datos
    - creamos el campo puntuacion con la siguiente formula: experiencia (años * 1.5) y sumando los puntos de catalán, excel y titulación(titulo)

```python
def calcular_puntuacion(datos):
    
    datos["puntuacion"] = (datos["experiencia"] * 1.5) + datos["catalan"] + datos["excel"] + datos["titulo"]
    
    return datos["puntuacion"] # no cal retornar ya que estamos manipulando el dicc datos directamente dentro de la funcion
```

- funcion classificar_candidato(recibe el diccionario de datos):
    - si la puntiacion es mayor o igual a 12:
        - asignamos su prioridad como Alta
    - si la puntiacion es menor a 12 y mayor o igual a 8:
        - asignamos su prioridad como Candidato Adecuado
    - si la puntiacion es menor a 18:
        - asignamos su prioridad como No Prioritario

```python
def clasificar_candidato(datos):
    if datos["puntuacion"] >= 12:
        datos["prioridad"] = "Prioridad Alta"
    elif 8 <= datos["puntuacion"] < 12:
        datos["prioridad"] = "Candidato Adecuado"
    else:
        datos["prioridad"] = "No Prioritario"

    return datos["prioridad"] # no cal retornar ya q estamos manipulando el dicc directamente dentro la funcion
```

- funcion mostrar_informe_final():

    - ordenamos el parametro recibido(diccionario candidatos_validos) según 
        1. el orden_prioridad,  ordena la lista de candidatos válidos siguiendo dos criterios a la vez. Primero los agrupa por su nivel de prioridad (de mayor a menor importancia), y si hay empate en la prioridad, los ordena de mayor a menor puntuación.

```python
def mostrar_informe_final(candidatos_validos):
    orden_prioridad = {"Prioridad Alta": 1, "Candidato Adecuado": 2, "No Prioritario": 3}
    
    candidatos_validos.sort(key=lambda x: (orden_prioridad[x["prioridad"]], -x["puntuacion"]))

    print("\nInforme final de candidatos:")
    for num, candidato in enumerate(candidatos_validos, start=1):
        print(f"{num}. Prioridad: {candidato['prioridad']}, Nombre: {candidato['nombre']}, Teléfono: {candidato['telefono']}, Puntuación: {candidato['puntuacion']}")
```

- Definir el diccionario "comportamientos" con los datos de segmento, mensaje y factor.

- Definir el diccionario "factores_edad" con los valores numéricos de cada rango de edad.


```python
comportamientos = {
    0: {
        "segmento": "Bajo interés",
        "mensaje": "Descubre nuestros productos",
        "factor": 0.10
    },
    1: {
        "segmento": "Interés inicial",
        "mensaje": "Tienes productos esperándote",
        "factor": 0.25
    },
    2: {
        "segmento": "Cerca de comprar",
        "mensaje": "Finaliza tu compra ahora",
        "factor": 0.50
    },
    3: {
        "segmento": "Abandono reciente",
        "mensaje": "Tu carrito sigue disponible",
        "factor": 0.75
    },
    4: {
        "segmento": "Abandono crítico",
        "mensaje": "Última oportunidad para completar tu compra",
        "factor": 0.90
    }
}

factores_edad = {
    "18-25": 1.20,
    "26-40": 1.00,
    "41-60": 0.85,
    "60+": 0.70
}
```

- Funció obtener_segment_i_factor(codigo):
    - Buscar en el diccionario "comportamientos" usando la clave "codigo".
    - Retornar los datos del comportamiento (segmento, mensaje y factor).

- Funció obtener_factor_edat(grupo_edad):
    - Buscar en el diccionario "factores_edad" usando la clave "grupo_edad".
    - Retornar el número del factor de edad encontrado.

- Funció calcular_risc(factor_codigo, factor_edad):
    - Calcular el resultado de multiplicar factor_codigo por factor_edad.
    - Retornar el riesgo final obtenido.

- Funció processar_clients(clientes):
    1. Inicializar una lista vacía llamada clientes_procesados.
    2. Para cada cliente dentro de la lista clientes:
        - Obtener "info_comp" llamando a obtener_segment_i_factor con su codigo_comportamiento.
        - Obtener "factor_edad" llamando a obtener_factor_edat con su grupo_edad.
        - Obtener "riesgo_final" llamando a calcular_risc con el factor e info_comp y el factor_edad.
        - Crear un nuevo diccionario llamado cliente_procesado uniendo todos los datos originales del cliente con los tres datos calculados (factor_codigo, factor_edad, riesgo_final, segmento y mensaje).
        - Añadir el nuevo diccionario cliente_procesado a la lista clientes_procesados.
    3. Retornar la lista clientes_procesados con la información de todos los clientes unificada.



```python
# 1. Creamos las funciones modulares
def obtener_segment_i_factor(codigo):
    return comportamientos[codigo]

def obtener_factor_edat(grupo_edad):
    return factores_edad[grupo_edad]

def calcular_risc(factor_codigo, factor_edad):
    return round(factor_codigo * factor_edad, 3)


# 2.  procesar_clientes usando las subfunciones de arriba
def processar_clients(clientes):     
    clientes_procesados = []     
    
    for cliente in clientes:                  
        clientes_procesados.append({             
            "id_cliente": cliente["id_cliente"],             
            "nombre": cliente["nombre"],             
            "edad": cliente["edad"],             
            "grupo_edad": cliente["grupo_edad"],             
            "codigo_comportamiento": cliente["codigo_comportamiento"],              

            # Colocamos las funciones directo en el diccionario sin crear variables previas:
            "factor_codigo": obtener_segment_i_factor(cliente["codigo_comportamiento"])["factor"],             
            "factor_edad": obtener_factor_edat(cliente["grupo_edad"]),             
            "riesgo_final": calcular_risc(
                obtener_segment_i_factor(cliente["codigo_comportamiento"])["factor"], 
                obtener_factor_edat(cliente["grupo_edad"])
            ),              
            "segmento": obtener_segment_i_factor(cliente["codigo_comportamiento"])["segmento"],             
            "mensaje": obtener_segment_i_factor(cliente["codigo_comportamiento"])["mensaje"]         
        })         
    return clientes_procesados


# procesar_clientes(clientes)
```

```python
import pandas as pd
```

```python

def crear_dataframe(clientes_procesados):
    df = pd.DataFrame(clientes_procesados).sort_values(by="riesgo_final", ascending=False)
    return df
```

```python
def mostrar_dataframe(df):
    print("\n======================= DATAFRAME FINAL DE CLIENTES =======================")
    print(df.to_string(index=False)) # Imprime la tabla limpia en la pantalla
```

```python
def mostrar_respostes(df):
  print("\n=== RESPUESTAS DE MARKETING ===")

  # Clientes con mayor riesgo, el df esta ya ordenado por riesgo
  primero = df.iloc[0]
  segundo = df.iloc[1]
  
  # 1. Quién contactar primero
  print(f"\n1. Contactar primero a: {primero['nombre']} ({primero['riesgo_final']}) y {segundo['nombre']} ({segundo['riesgo_final']}).")

  # 2. Diferencias por edad agrupando por comportamiento
  print("\n2. Comparación de edad y riesgo final dentro del mismo comportamiento:")
  # Agrupamos por comportamiento y mostramos la tabla ordenada por edad
  for grupo, codigo in df.groupby("codigo_comportamiento"):
    print(f"El grupo es: {grupo}") 
    # 1. Seleccionamos solo las columnas que queremos ver
    columnas_clave = codigo[["nombre", "edad", "grupo_edad", "riesgo_final"]]
    # 2. Imprimimos los resultados de este grupo en formato tabla limpia
    print(columnas_clave.to_string(index=False))
    
  # 3. ¿Qué clientes presentan tan poco interés que no vale la pena invertir presupuesto en ellos?
  print("\n3. ¿Qué clientes presentan tan poco interés que no vale la pena invertir presupuesto en ellos?")
  print(df[df["codigo_comportamiento"] == 0][["nombre", "segmento", "riesgo_final"]].to_string(index=False))

  # 4. ¿Qué cliente sería el más rentable de recuperar?
  # Extraemos de forma dinámica el cliente con mayor riesgo (primera fila)
  mas_rentable = df.iloc[0]

  print("\n4. ¿Qué cliente sería el más rentable de recuperar?")
  print(f"   El cliente más rentable de recuperar es {mas_rentable['nombre']} (Riesgo Final: {mas_rentable['riesgo_final']}).")
  print(f"   El mensaje publicitario ideal es: '{mas_rentable['mensaje']}'.")
```
