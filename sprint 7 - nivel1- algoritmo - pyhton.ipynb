Función obtener datos(): no recibe ningún parámetro
    -crea un diccionario datos con las siguientes claves que obtienen los valores de sus inputs
        -nombre
        -telefono
        -experiencia
        -catalan (niveles de 0 a 3)
        -excel (niveles de 0 a 3)

    retorna el diccionario datos

```python
def obtener_datos():
    datos = {
        "nombre": input("Ingresa el nombre: "),
        "telefono": input("Ingresa tu teléfono: "),
        "experiencia": input("Ingresa tus años de experiencia en la empresa: "),
        "catalan": input("""Conocimiento de Catalán:
            0 = Sin conocimiento 
            1 = Hasta nivel B1 
            2 = Nivel B2 
            3 = C1 o superior 
        """),
        "excel": input("""Dominio de Excel:
            0 = Nivel básico 
            1 = Uso de funciones 
            2 = Tablas dinámicas 
            3 = Visual Basic y Macros
        """),
        "titulo": input("""Titulada académica:
            0 = ESO 
            1 = FP 
            2 = Universidad 
            3 = Máster o superior
        """)
    }
    return datos
```

funcion validar_datos(datos): recibe el diccinario creado en obtener_datos
    intenta:
        delvolver la conversion de string a integer 
    si no lo consigue:
        imprime(el error)
        retorna False para que interrumpir la funcion de validar_datos
    
    si el campo nombre está vacío:
        print("El nombre no puede estar vacío")
        devuelve False para indicar que los datos no son válidos

    si el campo telefono no es un dígito o si no tiene 9 digitos:
        print("El teléfono debe ser un número de 9 dígitos")
        devuelve False para indicar que los datos no son válidos
    
    si despues de convertir a integer, el campo experiencia es negativo 
        print("Los años de experiencia no pueden ser negativos.")
        devuelve False para indicar que los datos no son válidos
    
    si despues de convertir a integer, el campo catalan  no es 0, 1, 2 o 3:
        print("El nivel de catalán debe ser 0, 1, 2 o 3.")
        devuelve False para indicar que los datos no son válidos

    si despues de convertir a integer,el campo excel no es 0, 1, 2 o 3:
        print("El dominio de Excel debe ser 0, 1, 2 o 3.")
        devuelve False para indicar que los datos no son válidos

    si despues de convertir a integer,el campo titulo no es 0, 1, 2, 3:
        print("El título académico debe ser 0, 1, 2 o 3.")
        devuelve False para indicar que los datos no son válidos

    devuelve True si todas las condiciones de validacion han sido exitosas

```python
def validar_datos(datos):
    try:
        datos["experiencia"] = float((datos["experiencia"]))
        datos["catalan"] = int((datos["catalan"]))
        datos["excel"] = int((datos["excel"]))
        datos["titulo"] = int((datos["titulo"]))
        # return int(val)
    except Exception as e:
        print("error = ", e)
        return False
    
    # Validar nombre
    if len(datos["nombre"].strip()) == 0:
        print("El nombre no puede estar vacío.")
        return False

    # Validar teléfono
    if datos["telefono"].isdigit() == False or len(datos["telefono"]) != 9:
        print("El teléfono debe ser un número de 9 dígitos.")
        return False

    # Validar años de experiencia
    if (datos["experiencia"]) < 0:
        print("Los años de experiencia no pueden ser negativos.")
        return False

    # Validar nivel de catalán
    if (datos["catalan"]) not in [0, 1, 2, 3]:
        print("El nivel de catalán debe ser 0, 1, 2 o 3.")
        return False

    # Validar dominio de Excel
    if (datos["excel"]) not in [0, 1, 2, 3]:
        print("El dominio de Excel debe ser 0, 1, 2 o 3.")
        return False

    # Validar título académico
    if (datos["titulo"]) not in [0, 1, 2, 3]:
        print("El título académico debe ser 0, 1, 2 o 3.")
        return False

    return True
```

funcion calcular_puntuacion(datos): recibe el diccionario datos
    creamos el campo puntuacion con la siguiente formula: experiencia (años * 1.5) y sumando los puntos de catalán, excel y titulación(titulo)

```python
def calcular_puntuacion(datos):
    
    datos["puntuacion"] = (datos["experiencia"] * 1.5) + datos["catalan"] + datos["excel"] + datos["titulo"]
    return datos["puntuacion"]
```

funcion classificar_candidato(recibe el diccionario de datos):
    si la puntiacion es mayor o igual a 12:
        asignamos su prioridad como Alta
    si la puntiacion es menor a 12 y mayor o igual a 8:
        asignamos su prioridad como Candidato Adecuado
    si la puntiacion es menor a 18:
        asignamos su prioridad como No Prioritario

```python
def clasificar_candidato(datos):
    if datos["puntuacion"] >= 12:
        datos["prioridad"] = "Prioridad Alta"
    elif 8 <= datos["puntuacion"] < 12:
        datos["prioridad"] = "Candidato Adecuado"
    else:
        datos["prioridad"] = "No Prioritario"
    return datos["prioridad"]
```

funcion mostrar_individual(recibe el diccionario datos):
    #Cada vez que alguien complete el formulario, se debe mostrar su puntuación individual mostrar_individual
    -nombre
    -telefono
    -años de experiencia
    -nivel de catalan (niveles de 0 a 3)
    -nivel de excel (niveles de 0 a 3)
    -titulación (ESO, FP, universidad)
    


```python
def mostrar_individual(datos):
    print(f"Nombre: {datos['nombre']}")
    print(f"Teléfono: {datos['telefono']}")
    print(f"Años de experiencia: {datos['experiencia']}")
    print(f"Nivel de catalán: {datos['catalan']}")
    print(f"Dominio de Excel: {datos['excel']}")
    print(f"Título académico: {datos['titulo']}")
    print(f"Puntuación total: {datos['puntuacion']}")
    print(f"Clasificación: {datos['prioridad']}")
    print("-" * 40)
```

funcion mostrar_informe_final(): recibe parametro una lista de diccionarios donde cada diccionario son los datos de cada candidato validado
    -declaramos orden_prioridad, diccionario con la puntiacion de cada tipo de prioridad
    -ordenamos el parametro recibido(diccionario candidatos_validos) según 1)el orden_prioridad, 2)puntuacion
    -mostramos la lista de diccionarios candidatos_validos(ordenada previamente por prioridad y puntuacion),


```python
def mostrar_informe_final(candidatos_validos):
    orden_prioridad = {"Prioridad Alta": 1, "Candidato Adecuado": 2, "No Prioritario": 3}
    
    candidatos_validos.sort(key=lambda x: (orden_prioridad[x["prioridad"]], -x["puntuacion"]))

    print("\nInforme final de candidatos:")
    for num, candidato in enumerate(candidatos_validos, start=1):
        print(f"{num}. Prioridad: {candidato['prioridad']}, Nombre: {candidato['nombre']}, Teléfono: {candidato['telefono']}, Puntuación: {candidato['puntuacion']}")
```

funcion main(): es donde se irán llamando a las funciones anteriores
    -declara lista vacia de candidatos_validos
    -mientras la longitud de candidatos_validos sea manor a 10:
        -crea la estructura datos que sera el diccionario de cada candidato ya que llama a obtener_datos()
        -si se valida esos datos:
            -se calcula la puntuacion de ese candidato
            -se clasifica a ese candidato segun prioridad
            -se muestra la info de ese candidato
            -se añade la info de ese candidato a la lista de candidatos_validos
    -muestra el informe final con la lista de diccinarios que represnta a los candidatos_validos

```python
def main():
    candidatos_validos = []
    while len(candidatos_validos) < 10:
        datos = obtener_datos()
        # validar_datos(datos)
        if validar_datos(datos):
            calcular_puntuacion(datos)
            clasificar_candidato(datos)
            mostrar_individual(datos)
            candidatos_validos.append(datos)
    mostrar_informe_final(candidatos_validos)
main()
```

```python
comportamientos = {
    0: {
        "segmento": "Bajo interés",
        "mensaje": "Descubre nuestros productos",
        "factor": 0.10
    },
    1: {
        "segmento": "Interés inicial",
        "mensaje": "Tienes productos esperándote",
        "factor": 0.25
    },
    2: {
        "segmento": "Cerca de comprar",
        "mensaje": "Finaliza tu compra ahora",
        "factor": 0.50
    },
    3: {
        "segmento": "Abandono reciente",
        "mensaje": "Tu carrito sigue disponible",
        "factor": 0.75
    },
    4: {
        "segmento": "Abandono crítico",
        "mensaje": "Última oportunidad para completar tu compra",
        "factor": 0.90
    }
}

factores_edad = {
    "18-25": 1.20,
    "26-40": 1.00,
    "41-60": 0.85,
    "60+": 0.70
}
```

funcion procesar_clientes(recibe como parámetro la lista diccionarios con los datos de los cliente):
    inicializamos la lista clientes_procesados donde añadiremos campos adicionales a la lista original de clientes
    
    iteramos sobre cada elemento de clientes
        calculamos los valores de los campos nuevos a añadir al diccionarios originales(sin ifs)
        -factor_codigo: Busca el código de comportamiento del cliente en la lista de comportamientos y extrae su factor numérico correspondiente
        -factor_edad:  Busca el grupo de edad del cliente en la lista general de edades para obtener su multiplicador de riesgo por edad correspondiente.
        -riesgo_final: Calcula la puntuación total de riesgo multiplicando el factor de su comportamiento por el factor de su edad
        -segmento : Identifica la categoría o etiqueta comercial (por ejemplo: "Alto Riesgo", "Estándar", "Premium") que le corresponde al cliente según su comportamiento
        -mensaje: Igual que en la línea anterior, acudimos al registro maestro de comportamientos usando el código del cliente, pero esta vez extrae el texto guardado en el apartado "mensaje".

        declaramos el diccionario clientes_procesado que tendrá los datos originales más los nuevos campos que acabamos de crear.

        anexamos cada cliente_procesado a la lista clientes_procesados
    retornamos la lista de clientes_procesados

```python
def procesar_clientes(clientes):     
    clientes_procesados = []     
    
    for cliente in clientes:         
        # print("cliente = ", cliente)         
        factor_codigo = comportamientos[cliente["codigo_comportamiento"]]["factor"]         
        factor_edad = factores_edad[cliente["grupo_edad"]]          
        riesgo_final = factor_codigo * factor_edad          
        segmento = comportamientos[cliente["codigo_comportamiento"]]["segmento"]         
        mensaje = comportamientos[cliente["codigo_comportamiento"]]["mensaje"]          
        cliente_procesado = {             
            # datos originales             
            "id_cliente": cliente["id_cliente"],             
            "nombre": cliente["nombre"],             
            "edad": cliente["edad"],             
            "grupo_edad": cliente["grupo_edad"],             
            "codigo_comportamiento": cliente["codigo_comportamiento"],              

            # datos nuevos             
            "factor_codigo": factor_codigo,             
            "factor_edad": factor_edad,             
            "riesgo_final": riesgo_final,              
            "segmento": segmento,             
            "mensaje": mensaje         
        }         
        clientes_procesados.append(cliente_procesado)         
        
    return clientes_procesados
```

primero importamos la librería de pandas

segundo definimos la funcion crear_dataframe que recibirá la lista de diccionarios clientes_procesados
    crea el dataframe en el que ordenada los clientes_procesados por riesgo_final en orden descendente

```python
import pandas as pd
def crear_dataframe(clientes_procesados):
    df = pd.DataFrame(clientes_procesados).sort_values(by="riesgo_final", ascending=False)
    return df
```

función principal main():
    declaramos clientes_procesados que obtendrá los datos de procesar_clientes(clientes)
    declaramos df cuyo valor será el dataframe despues de enviar la lista de clientes procesados a crear_dataframe
    retornamos el df para mostrarlo en el output

```python
def main():
    clientes_procesados = procesar_clientes(clientes)
    df = crear_dataframe(clientes_procesados)
    # print(df)
    return df
main()
```
