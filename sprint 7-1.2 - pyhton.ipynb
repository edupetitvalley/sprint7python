```python
comportamientos = {
    0: {
        "segmento": "Bajo interés",
        "mensaje": "Descubre nuestros productos",
        "factor": 0.10
    },
    1: {
        "segmento": "Interés inicial",
        "mensaje": "Tienes productos esperándote",
        "factor": 0.25
    },
    2: {
        "segmento": "Cerca de comprar",
        "mensaje": "Finaliza tu compra ahora",
        "factor": 0.50
    },
    3: {
        "segmento": "Abandono reciente",
        "mensaje": "Tu carrito sigue disponible",
        "factor": 0.75
    },
    4: {
        "segmento": "Abandono crítico",
        "mensaje": "Última oportunidad para completar tu compra",
        "factor": 0.90
    }
}

factores_edad = {
    "18-25": 1.20,
    "26-40": 1.00,
    "41-60": 0.85,
    "60+": 0.70
}

#   • » Fórmula del risc final 
#       risc_final = factor codi * factor edat
```

```python
def processar_clients(clients):
    clientes_procesados = []
    for client in clients:
        # print("client = ", client)
        factor_codi = comportamientos[client["codi_comportament"]]["factor"]
        factor_edat = factores_edad[client["grup_edat"]]

        riesgo_final = factor_codi * factor_edat

        segment = comportamientos[client["codi_comportament"]]["segmento"]
        missatge = comportamientos[client["codi_comportament"]]["mensaje"]

        client_procesado = {
            # datos originales
            "id_client": client["id_client"],
            "nom": client["nom"],
            "edat": client["edat"],
            "grup_edat": client["grup_edat"],
            "codi_comportament": client["codi_comportament"],

            # datos nuevos
            "factor_codi": factor_codi,
            "factor_edat": factor_edat,
            "risc_final": riesgo_final, 
            "segment": segment,
            "missatge": missatge
        }
        clientes_procesados.append(client_procesado)    
    return clientes_procesados

# processar_clients(clients)
```

```python
import pandas as pd
def mostrar_dataframe(clientes_procesados):
    df = pd.DataFrame(clientes_procesados).sort_values(by="risc_final", ascending=False)
    return df
```

```python
def main():
    clientes_procesados = processar_clients(clients)
    df = mostrar_dataframe(clientes_procesados)
    # print(df)
    return df
main()
```
