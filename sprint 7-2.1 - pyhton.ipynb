Exercici 1: Gestió de risc i priorització de pacients

    Treballes com a analista de dades en un centre mèdic especialitzat. Diversos pacients han enviat els seus símptomes a través d'un formulari online, però les dades arriben en un format desordenat i poc estructurat, i l'equip mèdic necessita que els organitzis per poder prioritzar l'ordre d'atenció segons la gravetat. 

A continuació tens una llista amb informació real procedent del sistema:

```python
import pandas as pd
```

🎯 Objectius

L'objectiu és transformar aquestes dades en informació útil per als metges. A partir dels registres en brut, has de:

    » Convertir cada string en una estructura neta (diccionari o similar).
    » Separar i normalitzar la llista de símptomes:
        minúscules
        eliminar espais
        dividir correctament per comes

    » Assignar un risc a cada símptoma utilitzant el diccionari nivells_risc.

    » Calcular el risc total de cada pacient sumant els valors.

    » Ordenar els pacients de major a menor risc per determinar la prioritat d'atenció.


```python
pacients = [ 
    "Maria|45|tos, Fiebre Alta, dolor_pecho", 
    "Luis|33|dolor_cabeza,faTiga", 
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos", 
    "Jordi|52|fatiga , tos", 
    "Anna|29|fiebre alta,dolor_cabeza" 
]

# El departament mèdic també t'ha proporcionat el següent diccionari de nivells de risc per símptoma:        

nivells_risc = { 
    "fiebre_alta": 3, 
    "dificultad_respirar": 5, 
    "dolor_pecho": 4, 
    "tos": 1, 
    "fatiga": 1, 
    "dolor_cabeza": 1 
}
```

```python
# pacients_dicc = [ 
#     {"id_client": 1, "nom": "Laura Serra", "edat": 23, "grup_edat": "18-25", "codi_comportament": 4}, 
```

```python
# pacientes tiene un formato de lista de srtings, cada string tiene el formato "nombre|edad|sintomas", donde los sintomas están separados por comas. La función llista_a_diccionari convierte esta lista de strings en una lista de diccionarios, donde cada diccionario representa un paciente con sus datos procesados.
def llista_a_diccionari(pacients):
    pacients_dicc = []
    for pacient in pacients:
        # print("pacient = ", pacient)
        paciente_procesado = {
            "nombre" : pacient.split("|")[0].strip(),
            "edad" : pacient.split("|")[1].strip(),
            "sintomas" : pacient.split("|")[2].strip().lower().replace(", ", ",").replace(" ,", ",").replace(" ", "_")
        }
        # print("paciente_procesado = ", paciente_procesado)
        pacients_dicc.append(paciente_procesado)    
    return pacients_dicc
```

```python
def calcular_risc(pacients_dicc):
    for pacient in pacients_dicc:
        pacient["riesgo_total"] = 0
        # print("pacient = ", pacient["sintomas"])
        lista_sintomas = pacient["sintomas"].split(",")
        # print("lista_sintomas = ", lista_sintomas)
        for sintoma in lista_sintomas:
            # print(sintoma, nivells_risc[sintoma])
            pacient["riesgo_total"] = pacient["riesgo_total"] + nivells_risc[sintoma]
            #quitamos el _ de los sintomas 
            pacient["sintomas"] = pacient["sintomas"].replace("_", " ")
    return pacients_dicc
```

```python
def mostrar_dataframe(clientes_procesados):
    df = pd.DataFrame(clientes_procesados)
    return df
    # print(df)
```

```python
def main():
    pacients_dicc = llista_a_diccionari(pacients)
    pacients_diccRiesgo = calcular_risc(pacients_dicc)
    return mostrar_dataframe(pacients_diccRiesgo).sort_values(by="riesgo_total", ascending=False)
main()
```
