```python
import pandas as pd 
import numpy as np
from numpy import random
# Para crear las "cajas" de caracteres en Python de una forma limpia y profesional, no hace falta que escribas todo el abecedario a mano en una lista. La forma estándar es utilizar una librería nativa de Python llamada #string, la cual ya viene con todas estas "cajas" prefabricadas de fábrica.
import string


# crear_contrasenya(12, majuscules=True, minuscules=True, numeros=True, signes=True)
# Output possible:
# WS6MwfRD:%B3

# majuscules=True, minuscules=True, numeros=True: Se definen con =True porque el enunciado exige que estos tres tipos de caracteres estén activados por defecto si el usuario no dice lo contrario.
def crear_contrasenya(longitud, majuscules=True, minuscules=True, numeros=True, signes=False):
    contrasenya = ""
    # para completar el resto de caracteres hasta alcanzar la longitud deseada, necesitas crear una "box" que junte todos los caracteres que el usuario ha activado (con True). Después, calculas cuántos caracteres te faltan por rellenar y los eliges al azar de esa gran caja.
    box = ""
    if majuscules:
        caixa_majuscules = string.ascii_uppercase
        box += caixa_majuscules
    if minuscules: 
        caixa_minuscules = string.ascii_lowercase
        box += caixa_minuscules
    if numeros:
        caixa_numeros = string.digits
        box += caixa_numeros
    if signes:
        caixa_signes = string.punctuation
        box += caixa_signes
    # print("box = ", box)


    for _ in range(longitud): #n Python, the underscore (_) is used as a throwaway variable name
        contrasenya += random_choice(box)     
    print("contrasenyaa = ", contrasenya)

def random_choice(box):
    # in order to use np.random need to turn string into a list ,To break a string down into individual single letters, you wrap the string in list()
    return np.random.choice(list(box))
    

def main():
    crear_contrasenya(12, signes=True)
main()
```
